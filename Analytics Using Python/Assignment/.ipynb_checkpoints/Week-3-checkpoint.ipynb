```python
import pandas as pd

#1.	Read the file Salaries.csv into a dataframe.
df = pd.read_csv("C:/Users/rupas/Desktop/Analytics Using Python/Assignment/Salaries.csv")
```

```python
#2.	We just want to use the first 100 rows of data for year 2014 of this dataset for further analysis. To do that, filter the data using a condition(.query) and then use loc/iloc to select 100 rows . Save that data in a new DataFrame and use the same for analysis.

data= df.query('Year == 2014').iloc[0:100]
data
```

```python
#3.	Plot a histogram for Total Pay column.

data.plot.hist(y='TotalPay',bins=8)
```

```python
#4.	Plot a barchart grouped by Jobtitle and mean values of BasePay.

data.groupby('JobTitle')['BasePay'].mean().plot.bar()
```

```python
#5.	Which employee has the highest TotalPay in the year 2014.
#Optional to plot it

highest_pay = data.sort_values('TotalPay', ascending=False).head(1)
highest_pay
```

```python
highest_pay.plot.bar()
```

```python
#6.	Pivot the data based on Year and JobTitle, with values as TotalPay. Plot the data with any chart of your choice.

pivot_data=data.pivot(columns=['Year','JobTitle'], values='TotalPay')
pivot_data.head()
```

```python
pivot_data.plot.line(marker='o',figsize=(10, 6))
```
