```python
###Load the data from csv file Salaries.csv in a dataframe. for eg :df = pd.read_csv('your_path/Salaries.csv')
import pandas as pd

salaries_dataset = pd.read_csv('D:/Webster University/Assignments/Spring 2024/Spring 1/Analytics Using Python/Assignment/Salaries.csv')
salaries_dataset
```

```python
### View the first few rows
salaries_dataset.head()
```

```python
###Explore the data using info(), describe()
```

```python
salaries_dataset.info()
salaries_dataset.describe()
```

```python
###What is the average BasePay

salaries_dataset['BasePay']=pd.to_numeric(salaries_dataset['BasePay'], errors='coerce')
average_basepay = salaries_dataset['BasePay'].mean()
average_basepay
```

```python
###What is the highest amount of OvertimePay in the dataset

salaries_dataset['OvertimePay']=pd.to_numeric(salaries_dataset['OvertimePay'], errors='coerce')
highest_overtime_pay = salaries_dataset['OvertimePay'].max()
highest_overtime_pay
```

```python
###How many unique job titles are there
salaries_dataset['JobTitle']=salaries_dataset['JobTitle'].astype('str')
unique_job_titles = salaries_dataset['JobTitle'].unique()
print("List of unique job titles: ", unique_job_titles)
print("Total unique job titles: ",salaries_dataset['JobTitle'].nunique())
```

```python
###Use the .query method to find rows where OvertimePay is greater than 1500. Display the result 

higher_overtimepay = salaries_dataset.query('OvertimePay > 1500')
higher_overtimepay
```

```python
# with .loc method
highest_overtime_pay = salaries_dataset.loc[salaries_dataset.OvertimePay > 1500]
highest_overtime_pay
```

```python
### Sort the data for each Year for the TotalPay in ascending order 
salaries_dataset.sort_values('TotalPay', ascending=True).head(5)
```

```python
###Which Job title has the highest TotalPay

highest_pay = salaries_dataset[['JobTitle','TotalPay']].max()
highest_pay
```

```python
###What is the average (mean) BasePay of all employees per year? (2011-2014
average_base_pay = salaries_dataset.groupby('Year')['BasePay'].mean()
average_base_pay
```
