```python
#Import Pandas, Matplotlib, Numpy, Statsmodels, Seaborn, and Scipy
import numpy as np
import pandas as pd
import statsmodels.api as sm
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Import your CSV file into Python as a Pandas dataframe
df = pd.read_csv('C:/Users/rupas/Desktop/Applied Business Statistics/4266261_Rupa_Shrestha_Assignment-2/dataset/Depression Student Dataset.csv')

#Print the top five rows and data types for each variable
print(df.head())

#Datatypes for each variable
print(df.dtypes)
```

```python
#display the summary statistics
df.describe()
```

```python
#Use Seaborn’s pairplot to create the big graphic containing all the different variables plotted against each other.
sns.pairplot(df)
```

```python
# Encoding categorical variables
df['Dietary_Habits'] = df['Dietary_Habits'].map({'Healthy': 2, 'Moderate': 1, 'Unhealthy': 0})
df['Have_you_ever_had_suicidal_thoughts_?'] = df['Have_you_ever_had_suicidal_thoughts_?'].map({'Yes': 1, 'No': 0})
df['Family_History_of_Mental_Illness'] = df['Family_History_of_Mental_Illness'].map({'Yes': 1, 'No': 0})
df['Depression'] = df['Depression'].map({'Yes': 1, 'No': 0})
sleep_mapping = {
    'Less than 5 hours': 1,
    '5-6 hours': 2,
    '7-8 hours': 3,
    'More than 8 hours': 4
}
df['Sleep_Duration'] = df['Sleep_Duration'].map(sleep_mapping)
```

```python
y = df['Depression']
xvars = ['Academic_Pressure','Study_Satisfaction','Sleep_Duration','Dietary_Habits', 'Have_you_ever_had_suicidal_thoughts_?', 'Study_Hours', 'Financial_Stress', 'Family_History_of_Mental_Illness']
X = df[xvars]


for i in xvars:
    sns.lmplot(x=i, y='Depression', fit_reg=True, data=df)
    plt.title(f'{i} vs Depression')
    plt.show()
```

```python
# Model 1: Using 'Study_Satisfaction' as the independent variable
model1 = sm.OLS.from_formula('Depression ~ Financial_Stress+Academic_Pressure', data=df)
results1 = model1.fit()
print(results1.summary())

```
